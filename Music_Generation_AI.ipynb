```python
from music21 import stream, note
import random

melody = stream.Stream()

notes = ['C4', 'D4', 'E4', 'F4', 'G4', 'A4', 'B4']

for i in range(20):
    melody.append(note.Note(random.choice(notes)))

melody.write('midi', fp='AI_Generated_Music.mid')

print("AI Generated Music created successfully!")
```
